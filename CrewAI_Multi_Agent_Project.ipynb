```python
import crewai

print("CrewAI installed successfully!")
print(crewai.__version__)
```

```python
from google.colab import userdata
import os
os.environ["OPENROUTER_API_KEY"] = userdata.get("OPENROUTER_API_KEY")

print("API Key loaded successfully!")
```

```python
from crewai import LLM
llm = LLM(
    model="openrouter/openai/gpt-4o-mini",
    api_key=os.environ["OPENROUTER_API_KEY"])
print("LLM is ready!")
```

```python
from crewai import Agent
researcher = Agent(
    role="AI Researcher",
    goal="Research and organize useful information about Artificial Intelligence in Education.",
    backstory="You are an expert AI researcher who collects and analyzes information carefully.",
    llm=llm,
    verbose=True)
print("Researcher Agent is ready!")
```

```python
writer = Agent(
    role="Report Writer",
    goal="Turn research findings into a clear, professional report.",
    backstory="You are a professional technical writer who creates organized and easy-to-understand reports.",
    llm=llm,
    verbose=True
)

print("Writer Agent is ready!")
```

```python
from crewai.tools import BaseTool

class TopicInformationTool(BaseTool):
    name: str = "Topic Information Tool"
    description: str = "Provides information about Artificial Intelligence in Education."

    def _run(self, topic: str) -> str:
        return """
        Artificial Intelligence in Education means using AI technologies
        to improve teaching and learning.

        Applications:
        - Personalized learning
        - Intelligent tutoring systems
        - Automated assessment
        - Learning analytics
        - Educational chatbots

        Benefits:
        - Personalized education
        - Faster feedback
        - Support for teachers
        - Automation of repetitive tasks

        Challenges:
        - Privacy concerns
        - Bias
        - Cost
        - Responsible use of AI
        """

topic_tool = TopicInformationTool()

print("Tool is ready!")
```

```python
researcher = Agent(
    role="AI Researcher",
    goal="Research and organize useful information about Artificial Intelligence in Education.",
    backstory="You are an expert AI researcher who collects and analyzes information carefully.",
    tools=[topic_tool],
    llm=llm,
    verbose=True
)
print("Researcher with Tool is ready!")
```

```python
writer = Agent(
    role="Report Writer",
    goal="Create a clear and professional report using the researcher's findings.",
    backstory="You are a professional writer who turns research into a well-organized report.",
    llm=llm,
    verbose=True
)

print("Writer Agent is ready!")
```

```python
from crewai import Task
research_task = Task(
    description="""
    Research the topic "Artificial Intelligence in Education".

    Use the Topic Information Tool to collect information about:
    1. What AI in education means.
    2. Main applications.
    3. Main benefits.
    4. Main challenges.

    Organize the findings clearly so another agent can use them.
    """,
    expected_output="""
    A structured research summary containing:
    - Definition
    - Applications
    - Benefits
    - Challenges
    """,
    agent=researcher
)
```

```python
writing_task = Task(
    description="""
    Using the researcher's findings, write a professional report
    about Artificial Intelligence in Education.

    The report should contain:
    - Introduction
    - Applications
    - Benefits
    - Challenges
    - Conclusion

    Make the report clear, organized, and easy to understand.
    """,
    expected_output="""
    A complete professional report about Artificial Intelligence in Education.
    """,
    agent=writer,
    context=[research_task]
)
```

```python
from crewai import Crew, Process
crew = Crew(
    agents=[researcher, writer],
    tasks=[research_task, writing_task],
    process=Process.sequential,
    verbose=True)
print("Crew is ready!")
```

```python
result = await crew.kickoff_async()
print("\n" + "=" * 60)
print("FINAL REPORT")
print("=" * 60)
print(result)
```
