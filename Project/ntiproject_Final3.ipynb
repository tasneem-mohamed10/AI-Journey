# 4.cleaning



```python
#drop dublicates
df1 = df1.drop_duplicates().reset_index(drop=True)
print(df1.duplicated().sum())
```

```python
#drop dublicates
df2 = df2.drop_duplicates().reset_index(drop=True)
print(df2.duplicated().sum())
```

```python
# fill Description
df1["Description"] = df1["Description"].fillna(
    df1.groupby("StockCode")["Description"].transform("first")
)
print(df1.isnull().sum())
```

```python
# fill null Description
df1["Description"] = df1["Description"].fillna("unknown")
print(df1.isnull().sum())
```

```python
# fill Description
df2["Description"] = df2["Description"].fillna(
    df2.groupby("StockCode")["Description"].transform("first")
)
print(df2.isnull().sum())
```

```python
# fill null Description
df2["Description"] = df2["Description"].fillna("unknown")
print(df2.isnull().sum())
```

```python
# handle customer id 
df1["Customer ID"] = df1["Customer ID"].fillna(-1)
print(df1.isnull().sum())
```

```python
# handle customer id 
df2["Customer ID"] = df2["Customer ID"].fillna(-1)
print(df2.isnull().sum())
```

```python
#make price zero=null
df1["Price"] = df1["Price"].replace(0,np.nan)
df2["Price"] = df2["Price"].replace(0,np.nan)
```

```python
print(len(df1[df1["Price"] == 0]))
print(len(df2[df2["Price"] == 0]))
```

```python
print(df1.isnull().sum())
print(df2.isnull().sum())
```

```python
# fill zero price
df1["Price"] = df1["Price"].fillna(
    df1.groupby(["StockCode","Description"])["Price"].transform("first")
)
df2["Price"] = df2["Price"].fillna(
    df2.groupby(["StockCode","Description"])["Price"].transform("first")
)
```

```python
print(df1.isnull().sum())
print(df2.isnull().sum())
```

```python
df1["Price"] = df1["Price"].fillna(0)
df2["Price"] = df2["Price"].fillna(0)
```

```python
print(df1.isnull().sum())
print(df2.isnull().sum())
```

```python
#remove inconsistent value
inconsistent1 = (df1["Quantity"] < 0) & (~df1["Invoice"].astype(str).str.startswith("C"))
df1 = df1[~inconsistent1].copy()
```

```python
#check  inconsistent value
print(len(df1[inconsistent1]))
```

```python
#remove inconsistent value(negative)
inconsistent2 = (df2["Quantity"] < 0) & (~df2["Invoice"].astype(str).str.startswith("C"))
df2 = df2[~inconsistent2].copy()
```

```python
#check inconsistent value(negative)
print(len(df2[inconsistent2]))
```

```python
#drop negative price
df1 = df1[df1["Price"] > 0].copy()
df2 = df2[df2["Price"] > 0].copy()
```

```python
#check negative price
df1[df1["Price"]<0]
df2[df2["Price"]<0]
```

```python
 #handle positive value in cancel
df1.loc[(df1["Invoice"].str.startswith("C", na=False)) & (df1["Quantity"] > 0), "Quantity"] *= -1
df2.loc[(df2["Invoice"].str.startswith("C", na=False)) & (df2["Quantity"] > 0), "Quantity"] *= -1
```

```python
#standarizing fields
df1["Country"] = df1["Country"].replace("EIRE", "Ireland")
df2["Country"] = df2["Country"].replace("EIRE", "Ireland")
df1["Country"] = df1["Country"].replace("RSA", "South Africa")
df2["Country"] = df2["Country"].replace("RSA", "South Africa")
df1["Country"] = df1["Country"].replace("USA", "United States")
df2["Country"] = df2["Country"].replace("USA", "United States")
```

```python
#separate dataframes from cancel
canceled1 = df1["Invoice"].astype(str).str.startswith("C")
canceled2 = df2["Invoice"].astype(str).str.startswith("C")

df1_canceled = df1[canceled1]
df2_canceled = df2[canceled2]

df1 = df1[~canceled1]
df2 = df2[~canceled2]
```

```python
#check dataframes from cancel
print(df1["Invoice"].astype(str).str.startswith("C").sum())
print(df2["Invoice"].astype(str).str.startswith("C").sum())
```

```python
#final shape
print(df1.shape)
print(df2.shape)
```

# integration


```python
df= pd.concat([df1, df2], ignore_index=True)
df_canceled=pd.concat([df1_canceled, df2_canceled], ignore_index=True)
```
